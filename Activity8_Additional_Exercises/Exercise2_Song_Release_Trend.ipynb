```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_top_1000_tracks.csv", encoding="utf-8")

print("Dataset loaded successfully!")
```

```python
student_name = "Correa"
student_id = "TUPM-26-1106"

color_bar = '#FFB6C1'
color_line = 'deeppink'

df['release_date'] = pd.to_datetime(
    df['release_date'],
    format='mixed',
    errors='coerce'
)

df['release_year'] = df['release_date'].dt.year

yearly_counts = df['release_year'].value_counts().sort_index()

cumulative_counts = yearly_counts.cumsum()

print("Yearly Song Releases:")
print(yearly_counts)

print("\nCumulative Song Releases:")
print(cumulative_counts)
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    yearly_counts.index,
    yearly_counts.values,
    color=color_bar,
    label='Yearly Song Count'
)

plt.plot(
    cumulative_counts.index,
    cumulative_counts.values,
    color='deeppink',
    marker='o',
    linewidth=2,
    label='Cumulative Song Count'
)

plt.title(
    f"{student_name} ({student_id})\nSong Release Trend"
)

plt.xlabel('Release Year')
plt.ylabel('Number of Songs')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
