```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("spotify_top_1000_tracks.csv", encoding="utf-8")

print("Dataset loaded successfully!")
print(df.head())
```

```python
student_name = "Correa"
student_id = "TUPM-26-1106"

id_num = int(student_id[-3:])

color_bar = '#FFB6C1'

print("Color Bar:", color_bar)
```

```python
longest_tracks = df.sort_values(
    by='duration_min',
    ascending=False
).head(10)

print(longest_tracks[['track_name', 'artist', 'duration_min']])
```

```python
plt.figure(figsize=(12, 6))

track_labels = (
    longest_tracks['track_name'] + ' - ' +
    longest_tracks['artist']
)

plt.barh(
    track_labels,
    longest_tracks['duration_min'],
    color=color_bar
)

plt.title(f"{student_name} ({student_id})\nTop 10 Longest Tracks")
plt.xlabel('Duration (min)')
plt.ylabel('Track Name - Artist')

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
