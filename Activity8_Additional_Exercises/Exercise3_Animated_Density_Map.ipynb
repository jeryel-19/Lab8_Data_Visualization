```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

df = pd.read_csv("spotify_top_1000_tracks.csv", encoding="utf-8")

print("Dataset loaded successfully!")
```

```python
student_name = "Correa"
student_id = "TUPM-26-1106"

cmap_color = 'plasma'

df['release_date'] = pd.to_datetime(
    df['release_date'],
    format='mixed',
    errors='coerce'
)

df['year'] = df['release_date'].dt.year

df = df.dropna(
    subset=['year', 'duration_min', 'popularity']
)

df['year'] = df['year'].astype(int)

years = sorted(df['year'].unique())

print("Available Years:", years)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

x_min = df['duration_min'].min()
x_max = df['duration_min'].max()
y_min = df['popularity'].min()
y_max = df['popularity'].max()

def update(year):
    ax.clear()

    yearly_data = df[df['year'] == year]

    hexbin = ax.hexbin(
        yearly_data['duration_min'],
        yearly_data['popularity'],
        gridsize=20,
        cmap=cmap_color,
        mincnt=1,
        extent=[x_min, x_max, y_min, y_max]
    )

    ax.set_title(
        f"{student_name} ({student_id})\n"
        f"Animated Density Map - Year {year}"
    )

    ax.set_xlabel('Duration (min)')
    ax.set_ylabel('Popularity')

    return hexbin,

animation = FuncAnimation(
    fig,
    update,
    frames=years,
    interval=800,
    repeat=True
)

plt.close(fig)

display(HTML(animation.to_jshtml()))
```
